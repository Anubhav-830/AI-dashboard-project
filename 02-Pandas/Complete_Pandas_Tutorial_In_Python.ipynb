list vs numpy- Time taken

```python
from time import process_time
```

Time taken by list

```python
python_list=[i for i in range(100000000)]
start_time=process_time()
python_list=[i+5 for i in python_list]    #as we can see it take a lot of time to
end_time=process_time()                   #execute the output so for this we have a scope of numpy library
print(end_time-start_time)
```

Take numpy to fid time

```python
np_array=np.array([i for i in range(100000000)])
start_time=process_time()
np_array += 5
end_time=process_time()
print(end_time-start_time)
```
